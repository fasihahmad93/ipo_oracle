```python
import logging
import os
import sys
from pathlib import Path

project_root = Path.cwd().resolve()

# Walk upward until we find the repo root, which contains the src/ directory.
for candidate in (project_root, *project_root.parents):
    if (candidate / 'src').exists():
        project_root = candidate
        break
else:
    raise FileNotFoundError(
        f"Could not find the project root from {Path.cwd().resolve()}. "
        "Ensure the notebook is opened inside the repository or a child directory."
    )

src_directory = project_root / 'src'
llm_orchestration_directory = src_directory / 'llm_orchestration'
web_crawling_directory = src_directory / 'web_crawling'

for directory in [str(src_directory), str(web_crawling_directory), str(llm_orchestration_directory)]:
    if directory not in sys.path:
        sys.path.insert(0, directory)

logging.basicConfig(level=logging.INFO, format='%(asctime)s | %(levelname)s | %(name)s | %(message)s')

print(f'Project root: {project_root}')
print(f'Web crawling directory: {web_crawling_directory}')
```
